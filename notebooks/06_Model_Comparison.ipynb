```python
from pathlib import Path
import sys

from IPython.display import Image as NotebookImage, display

project_root = next(
    (folder for folder in (Path.cwd(), *Path.cwd().parents) if (folder / "src" / "config.py").is_file()),
    None,
)
if project_root is None:
    raise FileNotFoundError("Chuyển thư mục làm việc về dự án trước. Trên Colab, xem hướng dẫn trong README.")
sys.path.insert(0, str(project_root))

from src.config import Paths
paths = Paths(project_root)
```

```python
import pandas as pd

from src.compare_models import compare_models
from src.evaluate import evaluate_models

metrics = evaluate_models(paths)
comparison = compare_models(paths)
display(pd.DataFrame(comparison["rows"]))
print(comparison["recommendation"])
display(NotebookImage(filename=str(paths.figures / "model_comparison.png")))
```

```python
for key in comparison["model_keys"]:
    print(key)
    display(NotebookImage(filename=str(paths.figures / f"{key}_confusion_matrix.png")))
    pairs = pd.read_csv(paths.metrics / f"{key}_confusion_pairs.csv")
    display(pairs.head(10))
    gallery = paths.figures / f"{key}_mistakes.png"
    if gallery.is_file():
        display(NotebookImage(filename=str(gallery)))
```
